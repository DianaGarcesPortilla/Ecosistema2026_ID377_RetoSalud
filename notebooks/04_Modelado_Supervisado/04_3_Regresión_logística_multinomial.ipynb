
**Objetivo:**

Entrenar un modelo de Regresión Logística Multinomial para
predecir el nivel de peligro asociado a casos de violencia
 intrafamiliar y de pareja.

Se evaluarán cuatro escenarios:

- Original
- Balanceado_500
- Balanceado_1000
- Balanceado_2000

El seguimiento experimental se realizará mediante MLflow.


```python

#Importar librerías

#Generales
import os
import json
import tempfile
import time

import pandas as pd
import numpy as np

#MLflow
import mlflow
import mlflow.sklearn
from mlflow.models import infer_signature


#Scikit-learn
import sklearn
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LogisticRegression


from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    fbeta_score
)
```

```python
#Configuración MLflow

#Experimento MLflow

NOMBRE_EXPERIMENTO = (
    "/Shared/Prediccion_Nivel_Peligro"
)

mlflow.set_experiment(
    NOMBRE_EXPERIMENTO
)
```

```python
#Cargar datasets

# Ruta Delta

ruta = (
    "ml_proyecto_7405607705157039.default"
)


# Dataset entrenamiento

train_original = spark.table(
    f"{ruta}.train_original"
)


train_balanceado_500 = spark.table(
    f"{ruta}.train_balanceado_500"
)


train_balanceado_1000 = spark.table(
    f"{ruta}.train_balanceado_1000"
)


train_balanceado_2000 = spark.table(
    f"{ruta}.train_balanceado_2000"
)

# Dataset prueba

test = spark.table(
    f"{ruta}.test"
)
```

```python
#Cargar pesos de clase

# Tabla pesos

df_pesos = spark.table(
    f"{ruta}.pesos_clases_modelos"
)
```

```python
#Crear diccionario de pesos por escenario

pesos_escenarios = {}

escenarios_pesos = (
    df_pesos
    .select("escenario")
    .distinct()
    .toPandas()["escenario"]
)

for escenario in escenarios_pesos:

    datos = (
        df_pesos
        .filter(
            df_pesos.escenario == escenario
        )
        .toPandas()
    )

    pesos_escenarios[escenario] = dict(
        zip(
            datos["clase"],
            datos["peso_clase"]
        )
    )
```

```python
#Variables del modelo

variables_modelo = [

    "sexo_victima_cod",
    "ciclo_vital_cod",
    "escolaridad_cod",
    "estado_civil_cod",
    "contexto_del_hecho_cod",
    "mecanismo_causal_cod",
    "presunto_agresor_cod"

]

target = "nivel_peligro"
```

```python
#Diccionario de escenarios

datasets = {

    "original":
        train_original,

    "balanceado_500":
        train_balanceado_500,

    "balanceado_1000":
        train_balanceado_1000,

    "balanceado_2000":
        train_balanceado_2000
}
```

```python
#Validación de estructura

columnas_requeridas = (
    variables_modelo + [target]
)


for nombre, df in datasets.items():

    faltantes = (
        set(columnas_requeridas)
        -
        set(df.columns)
    )

    if faltantes:

        raise ValueError(
            f"{nombre} tiene columnas faltantes: {faltantes}"
        )

faltantes_test = (
    set(columnas_requeridas)
    -
    set(test.columns)
)

if faltantes_test:

    raise ValueError(
        f"test tiene columnas faltantes: {faltantes_test}"
    )

print("Validación de columnas correcta")
```

```python
# Parámetros del modelo de Regresión Logística multinomial

PARAMETROS_LR = {

    "solver": "lbfgs",
    "max_iter": 1000,
    "random_state": 42,
    "multi_class": "multinomial"
}
```

```python
#Función construir_pipeline()

def construir_pipeline(
    variables_modelo,
    pesos_clase
):

    preprocesador = ColumnTransformer(

        transformers=[

            (
                "categoricas",

                OneHotEncoder(
                    handle_unknown="ignore"
                ),

                variables_modelo
            )
        ]

    )


    modelo = LogisticRegression(

        class_weight=pesos_clase,

        **PARAMETROS_LR
    )


    pipeline = Pipeline(

        steps=[

            (
                "encoder",
                preprocesador
            ),

            (
                "modelo",
                modelo
            )

        ]

    )


    return pipeline
```

```python
#Función calcular métricas

def calcular_metricas(
    y_real,
    y_pred
):

    reporte = classification_report(

        y_real,
        y_pred,
        output_dict=True,
        zero_division=0

    )

def calcular_metricas(
    y_real,
    y_pred
):

    from sklearn.metrics import (
        classification_report,
        confusion_matrix,
        accuracy_score,
        precision_score,
        recall_score,
        f1_score,
        fbeta_score
    )

    # Reporte por clase
    reporte = classification_report(
        y_real,
        y_pred,
        output_dict=True,
        zero_division=0
    )

    # Matriz de confusión
    clases = [
    "Peligro bajo",
    "Peligro moderado",
    "Peligro grave",
    "Peligro extremo"
    ]

    matriz = confusion_matrix(
        y_real,
        y_pred,
        labels=clases
    )

    # Especificidad para Peligro extremo
    
    idx = clases.index("Peligro extremo")

    TP = matriz[idx, idx]

    FN = matriz[idx, :].sum() - TP

    FP = matriz[:, idx].sum() - TP

    TN = matriz.sum() - TP - FN - FP

    specificity_extremo = (
        TN / (TN + FP)
        if (TN + FP) > 0
        else 0
    )

    # Diccionario de métricas
    
    metricas = {

        # Métricas globales
        
        "accuracy":

            accuracy_score(
                y_real,
                y_pred
            ),

        "precision_macro":

            precision_score(
                y_real,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "recall_macro":

            recall_score(
                y_real,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "f1_macro":

            f1_score(
                y_real,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "f2_macro":

            fbeta_score(
                y_real,
                y_pred,
                beta=2,
                average="macro",
                zero_division=0
            ),

        Clase Peligro extremo
        
        "precision_extremo":

            reporte["Peligro extremo"]["precision"],

        "recall_extremo":

            reporte["Peligro extremo"]["recall"],

        "f1_extremo":

            reporte["Peligro extremo"]["f1-score"],

        "f2_extremo":

            fbeta_score(
                y_real,
                y_pred,
                labels=["Peligro extremo"],
                average="macro",
                beta=2,
                zero_division=0
            ),

        "specificity_extremo":

            specificity_extremo,

        "support_extremo":

            reporte["Peligro extremo"]["support"]

    }

    return metricas, reporte, matriz
```

```python
#Función registrar_mlflow()

# Registro MLflow

def registrar_mlflow(

    pipeline,

    escenario,

    metricas,

    reporte,

    matriz,

    tiempo_entrenamiento,

    X_train,

    X_test

):

    
    # Parámetros
    
    mlflow.log_param(
        "modelo",
        "Regresion_Logistica"
    )


    mlflow.log_param(
        "escenario",
        escenario
    )


    mlflow.log_param(
        "numero_variables",
        len(variables_modelo)
    )


    mlflow.log_param(
        "n_train",
        len(X_train)
    )


    mlflow.log_param(
        "n_test",
        len(X_test)
    )


    mlflow.log_param(
        "sklearn_version",
        sklearn.__version__
    )


    mlflow.log_params(
        PARAMETROS_LR
    )


  
    # Métricas
   
    for nombre, valor in metricas.items():

        mlflow.log_metric(

            nombre,

            float(valor)

        )

    mlflow.log_metric(

        "tiempo_entrenamiento",

        tiempo_entrenamiento

    )

  
    # Signature

    input_example = X_train.head(5)

    signature = infer_signature(

        input_example,

        pipeline.predict(
            input_example
        )

    )

    # Modelo MLflow

    mlflow.sklearn.log_model(

        sk_model=pipeline,

        artifact_path="modelo",

        signature=signature,

        input_example=input_example

    )

    # Artefactos

    with tempfile.TemporaryDirectory() as tmp:


        reporte_path = os.path.join(

            tmp,

            "classification_report.json"

        )


        with open(

            reporte_path,

            "w",

            encoding="utf-8"

        ) as archivo:


            json.dump(

                reporte,

                archivo,

                indent=4,

                ensure_ascii=False

            )


        mlflow.log_artifact(

            reporte_path

        )



        matriz_path = os.path.join(

            tmp,

            "matriz_confusion.csv"

        )


        pd.DataFrame(
            matriz
        ).to_csv(

            matriz_path,

            index=False

        )


        mlflow.log_artifact(

            matriz_path

        )
```

**Función entrenar_regresion_logistica()**

Se une:

-Convertir los DataFrame de Spark train_df, test_df, en DataFrame de pandas y se almacenan en train_pd y test_pd, respectivamente.

-Pipeline.

-Entrenamiento.

-Predicción.

-Métricas.

-MLflow Run.

-run_id.

-Resultado para la tabla final.

Después se ejecutaran los 4 escenarios:

Original.
Balanceado_500.
Balanceado_1000.
Balanceado_2000.

```python
#Función entrenar_regresion_logistica()

# Entrenamiento Regresión Logística

def entrenar_regresion_logistica(

    escenario,

    train_df,

    test_df,

    variables_modelo,

    target,

    pesos_clase

):

    """
    Entrena, evalúa y registra un modelo
    de Regresión Logística Multinomial
    para un escenario específico.
    """
    
    # Spark -> Pandas
    
    train_pd = (
        train_df
        .select(
            variables_modelo + [target]
        )
        .toPandas()
    )


    test_pd = (
        test_df
        .select(
            variables_modelo + [target]
        )
        .toPandas()
    )

    # Separar variables
    
    X_train = train_pd[
        variables_modelo
    ]


    y_train = train_pd[
        target
    ]


    X_test = test_pd[
        variables_modelo
    ]


    y_test = test_pd[
        target
    ]

  
    # Crear Pipeline

    pipeline = construir_pipeline(

        variables_modelo,

        pesos_clase

    )

  
    # Run MLflow

    with mlflow.start_run(

        run_name=f"LR_{escenario}"

    ) as run:


        run_id = run.info.run_id

        # Entrenamiento
        
        inicio = time.time()


        pipeline.fit(

            X_train,

            y_train

        )


        tiempo_entrenamiento = (
            time.time()
            -
            inicio
        )

        
        # Predicción
        
        y_pred = pipeline.predict(

            X_test

        )


        # Métricas
        
        metricas, reporte, matriz = calcular_metricas(

            y_test,

            y_pred

        )


        metricas[
            "tiempo_entrenamiento"
        ] = tiempo_entrenamiento


        # Registrar MLflow
        
        registrar_mlflow(

            pipeline=pipeline,

            escenario=escenario,

            metricas=metricas,

            reporte=reporte,

            matriz=matriz,

            tiempo_entrenamiento=tiempo_entrenamiento,

            X_train=X_train,

            X_test=X_test

        )

    # Resultado para tabla Delta
    
    resultado = {


        "run_id":

            run_id,


        "modelo":

            "Regresion_Logistica",


        "escenario":

            escenario,


        **metricas

    }


    return resultado
```

```python
#Ejecutar los cuatro escenarios

# Entrenamiento escenarios

resultados = []


for escenario, train_df in datasets.items():

    print(
        "=" * 60
    )

    print(
        f"Entrenando escenario: {escenario}"
    )

    print(
        "=" * 60
    )


    resultado = entrenar_regresion_logistica(

        escenario=escenario,

        train_df=train_df,

        test_df=test,

        variables_modelo=variables_modelo,

        target=target,

        pesos_clase=pesos_escenarios[escenario]

    )


    resultados.append(
        resultado
    )


print(
    "Entrenamiento finalizado"
)
```

```python
#Crear tabla de resultados

# Tabla métricas

df_metricas = pd.DataFrame(
    resultados
)


display(
    df_metricas
)
```

```python
# Guardar resultados

df_metricas_spark = spark.createDataFrame(
    df_metricas
)

(
    df_metricas_spark
    .write
    .format("delta")
    .mode("overwrite")
    .option(
        "overwriteSchema",
        "true"
    )
    .saveAsTable(
        f"{ruta}.metricas_regresion_logistica"
    )
)
```
